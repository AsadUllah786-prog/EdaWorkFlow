### Read the CSV
I usually start with the raw data before changing anything.

```python
from pathlib import Path
import pandas as pd

# Works if I open Jupyter from the repo root or from notebooks/
data_path = Path("data/sales_raw.csv")
if not data_path.exists():
    data_path = Path("../data/sales_raw.csv")

df = pd.read_csv(data_path)
df.head(3)
```

```python
print("Rows, columns:", df.shape)
print("File:", data_path.name)
print(df.columns.tolist())
```

### Loading only the columns I need
For a quick question, I do not always need all 15 columns.

```python
mini = pd.read_csv(
    data_path,
    usecols=["order_id", "order_date", "city", "revenue_pkr"],
    dtype={"order_id": "string"},
    na_values=["NA", "N/A", "not available"]
)
mini.head()
```

### Excel and JSON practice
These are sample extracts, not separate sources of truth.

```python
data_dir = data_path.parent
excel_df = pd.read_excel(data_dir / "sales_sample.xlsx")
json_df = pd.read_json(data_dir / "sales_sample.json")
print("Excel sample:", excel_df.shape)
print("JSON sample:", json_df.shape)
json_df[["order_id", "product", "revenue_pkr"]].head()
```

### One loading mistake to watch for
Order IDs should stay as labels, not numerical measurements. Dates also need explicit parsing rules later.

```python
# Peek at the raw text before deciding the date format
print(df["order_date"].head(8).tolist())
print("Missing order dates:", df["order_date"].isna().sum())
```
